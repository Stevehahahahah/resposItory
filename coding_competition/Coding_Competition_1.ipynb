## Problem 1: Sum of Even Numbers 

Write a Python script that uses a **function** and a **loop** to calculate the sum of all even numbers from `1` through `n`. 

### Requirements 
Create a function named `sum_even(n)`. The function should: 
- Accept an integer `n` as a parameter. 
- Use a loop to find all even numbers from `1` through `n`. 
- Return the sum of those even numbers. Your program should ask the user to enter an integer value for `n`. 
- If the user enters a value less than `-1`, print: `Invalid input` 
- Otherwise, call `sum_even(n)` and print the result in the following format: `Sum of even numbers: 30` 

### Example 
If the user enters: `10` the program should calculate: `2 + 4 + 6 + 8 + 10 = 30` and print: `Sum of even numbers: 30`

````python
def sum_even(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i
    return total


n = int(input("Enter an integer n: "))

if n < -1:
    print("Invalid input")
else:
    print(f"Sum of even numbers: {sum_even(n)}")
````

## Problem 2: Even Number Square 

Write a Python function named `even_square(n)` that takes an integer `n` and prints an `n × n` square. 

### Requirements The function should: 
- Accept an integer `n` as a parameter. 
- Create an `n × n` grid using the numbers from `1` through `n²`. 
- Print each **even number** as itself. 
- Replace each **odd number** with an asterisk `*`. 
- Print the values row by row to form a square. 
- Include a space between each printed value. 

### Example 
If `n = 3`, the numbers in the square would normally be: 
```python
1 2 3
4 5 6
7 8 9 
```

After replacing the odd numbers with `*`, the function should print: 
``` python
* 2 *
4 * 6 
* 8 *
```

````python
def even_square(n):
    num = 1
    for row in range(n):
        values = []
        for col in range(n):
            if num % 2 == 0:
                values.append(str(num))
            else:
                values.append("*")
            num += 1
        print(" ".join(values))


even_square(3)
````

## Problem 3: Hollow Diamond

Write a Python script that draws a **hollow diamond** inside an `n × n` square grid.

### Requirements

Your program should:

- Ask the user to enter an integer `n`.
- Use an asterisk `*` to draw the edge of the diamond.
- Use a dot `.` for all other positions in the grid.
- Print a space between each character.
- Continue asking for values until the user enters `-1`.
- If the value of `n` cannot produce a centered diamond, print:

  `Invalid input`

A valid diamond requires `n` to be an odd number greater than or equal to `3`.

### Example

If the user enters:

`5`

the program should print:

```python
. . * . .
. * . * .
* . . . *
. * . * .
. . * . .
```

````python
def hollow_diamond(n):
    center = n // 2
    for row in range(n):
        line = []
        for col in range(n):
            # A cell is on the diamond's edge when its distance from the center equals the radius
            if abs(row - center) + abs(col - center) == center:
                line.append("*")
            else:
                line.append(".")
        print(" ".join(line))


while True:
    try:
        n = int(input("Enter an integer n (-1 to quit): "))
    except ValueError:
        print("Invalid input")
        continue

    if n == -1:
        break

    if n < 3 or n % 2 == 0:
        print("Invalid input")
    else:
        hollow_diamond(n)
````
